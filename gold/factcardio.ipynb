```sql
USE CATALOG pf1;
```

```sql

CREATE OR REPLACE VIEW gold.factcardio AS
SELECT
    id                  AS PatientId,
    age_years           AS AgeYears,
    age_group_id        AS IdAgeGroup,
    gender              AS IdGender,
    height_cm           AS HeightCm,
    weight_kg           AS WeightKg,
    bmi                 AS BMI,
    ap_hi               AS SystolicBP,
    ap_lo               AS DiastolicBP,
    pulse_pressure      AS PulsePressure,
    hypertension        AS HasHypertension,
    cholesterol         AS IdCholesterolType,
    gluc                AS IdGlucoseType,
    smoke               AS IsSmoker,
    alco                AS DrinksAlcohol,
    active              AS IsPhysicallyActive,
    cardio              AS HasCardiovascularDisease,
    _silverIngestTime   AS SilverIngestTime
FROM silver.cardiosilver
WHERE is_current = TRUE
  AND YEAR(_silverIngestTime) >= YEAR(CURRENT_DATE) - 2;
```

```sql
COMMENT ON VIEW gold.factcardio IS 
'Cardiovascular fact view. Contains only current SCD-2 records (is_current=TRUE) 
from the last 3 years (current year and 2 previous years). 
Source: silver.cardiosilver';
```
